# Figure per la tesi
Rigenerazione delle figure a partire da metriche e predizioni persistite.

```python
import json
import os
from pathlib import Path

import pandas as pd

from tesi_qml.evaluation.plots import plot_confusion_matrix, plot_learning_curves, save_figure

PROJECT_ROOT = Path(os.environ.get("TESI_PROJECT_ROOT", Path.cwd())).resolve()
if not (PROJECT_ROOT / "configs").is_dir() and (PROJECT_ROOT.parent / "configs").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
run_dir = next((PROJECT_ROOT / "results/runs").rglob("test_metrics.json")).parent
metrics = json.loads((run_dir / "test_metrics.json").read_text(encoding="utf-8"))
```

## Confusion matrix

```python
figure = plot_confusion_matrix(
    metrics["confusion_matrix"]["raw"], ("elliptical", "merger", "spiral")
)
figure
```

## Curve di apprendimento

```python
history = pd.read_json(run_dir / "history.jsonl", lines=True)
plot_learning_curves(history)
```

## Esportazione PNG e PDF
Entrambi i formati sono prodotti dalla stessa figura e dallo stesso stile.

```python
output = PROJECT_ROOT / "results/thesis_figures/confusion_matrix_example"
save_figure(figure, output)
```
