# Audit del dataset
Controlli descrittivi sul manifest già generato. Il notebook non modifica i dati.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path(os.environ.get("TESI_PROJECT_ROOT", Path.cwd())).resolve()
if not (PROJECT_ROOT / "configs").is_dir() and (PROJECT_ROOT.parent / "configs").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
manifest = pd.read_parquet(PROJECT_ROOT / "data/manifests/final.parquet")
manifest.head()
```

## Distribuzione delle classi
Il bilanciamento viene verificato sul manifest finale, non contando i file nella cartella immagini.

```python
counts = manifest["label"].value_counts().sort_index()
display(counts.to_frame("n"))
counts.plot.bar(
    color=["#3366CC", "#DC3912", "#109618"], ylabel="Campioni", title="Manifest finale per classe"
)
plt.tight_layout()
```

## Controlli di qualità e leakage

```python
audit = {
    "rows": len(manifest),
    "unique_objid": manifest["objid"].astype(str).nunique(),
    "unique_hashes": manifest["sha256"].nunique(),
    "missing_values": int(manifest.isna().sum().sum()),
}
pd.Series(audit, name="value")
```
